<h2 style="font-size:21px; font-weight:600; color:#222; margin-bottom:6px;">
Definitions
</h2>

<p style="font-size:16px; line-height:1.65; color:#444; margin-top:0; margin-bottom:22px;">
The following definitions establish the terminology used throughout the exercises below.
</p>

</div>

<div style="max-width:950px; margin:0 auto; font-family:Georgia, 'Times New Roman', serif; font-size:16px; line-height:1.65; color:#222;">

<p style="margin-bottom:18px;">
<strong>Effect.</strong> An effect is something that changes the distribution of possible experimental outcomes across repetitions—not something that appears when <i>p</i> &lt; 0.05. An effect represents a shift in the data-generating process that moves the sampling distribution away from what we would expect under the null hypothesis.
</p>

<p style="margin-bottom:18px;">
<strong>Sample size (<i>n</i>).</strong> The number of observations used to calculate a statistic within each sample. For example, if leaf length is measured on 30 leaves within each sampled plot, the sample size is <i>n</i> = 30.
</p>

<p style="margin-bottom:18px;">
<strong>Number of simulations.</strong> The number of independent samples generated from the population. Each simulated sample represents a new realization of the sampling process and produces its own value of the statistic of interest. In a field setting, an analogous design might involve sampling multiple independent plots or locations, with each plot or location providing a separate realization of the local population. Whether such units can be treated as independent samples depends on the sampling design and the scientific question.
</p>

<p style="margin-bottom:18px;">
<strong>&mu;.</strong> Population mean.
</p>

<p style="margin-bottom:18px;">
<strong>&sigma;.</strong> Population standard deviation.
</p>

</div>

<div style="max-width:950px; margin:32px auto 0 auto; font-family:Georgia, 'Times New Roman', serif;">

<h2 style="font-size:21px; font-weight:600; color:#222; margin-bottom:8px;">
Sample statistics as random variables
</h2>

<p style="font-size:16px; line-height:1.65; color:#333; margin-top:0;">
A <strong>sample statistic</strong> is a numerical quantity computed from observed data, such as a mean or standard deviation. Before data are collected, we do not know which sample will be obtained and therefore do not know the value the statistic will take. A statistic is consequently a <em>random variable before observation</em> and a numerical value after observation.
</p>

</div>

```python
# Import the needed libraries
import numpy as np
from numpy.random import default_rng
from scipy.stats import norm, t
import matplotlib.pyplot as plt
from __future__ import print_function

rng = np.random.default_rng()

# Population parameters
mu0 = 0.0          # Null mean
mu1 = 0.0          # Alternative mean
sigma = 0.5        # Population standard deviation

# Simulation parameters
sample_size = 30   # Observations per sample
n_sims = 1000      # Number of independent simulated samples
alpha = 0.05       # Significance level
```

Simulation step

<h3 style="font-size:18px; font-weight:600; color:#333; margin-top:0; margin-bottom:6px;">
Generate the sampling distribution
</h3>

<p style="font-size:15px; line-height:1.6; color:#555; margin-top:0; margin-bottom:0;">
The code below repeatedly draws random samples from a normally distributed population. A sample mean is calculated for each independent sample, producing a sampling distribution of the mean.
</p>

</div>

```python
# Calculate the sample mean for each randomly generated sample

def sampling_and_power(mu=mu0, sigma=sigma, sample_size=sample_size, n_sims=n_sims):
    
    means = []
    z_stats = []
    p_vals = []
    
    for _ in range(n_sims):
        
        # Create a psuedo-random array by sampling from a normal distribution
        sample = rng.normal(mu, sigma, sample_size)
        # Calculate the mean of the pseudo-random array
        xbar = sample.mean()
        # The test statistic measures extremeness relative to variability
        z = (xbar - mu0) / (sigma / np.sqrt(sample_size))
        # Calculate the theoretical p-value
        p = 2 * (1 - norm.cdf(abs(z)))
        # We could also calculate an empirical p-value
        # p_empirical = np.mean(abs(z_stats) >= abs(z_obs))
        
        # Store values
        means.append(xbar)
        z_stats.append(z)
        p_vals.append(p)

    p_vals = np.array(p_vals)
    # Calculate an alternative z-statistic
    z_alt = (mu - mu0) / (sigma / np.sqrt(sample_size))
    # Calculate the statistical power of the simulations
    power = np.mean(p_vals < alpha)
    
    # Create the fgiure environment
    fig, axes = plt.subplots(1, 3, figsize=(15,4))
    
    # LEFT-HAND PLOT
    # Identify one specific experiment to plot
    n = int(0.5 * n_sims)
    obs_mean = means[n]
    # Alternative hypothesis sample means histogram
    axes[0].hist(means, bins=50, density=True, color="forestgreen", alpha=0.7, label="Simulated sample means")
    # Null sampling distribution (theoretical)
    se = sigma / np.sqrt(sample_size)
    # Create a null distribution
    x = np.linspace(mu0 - 4*se, mu0 + 4*se, 400)
    null_pdf = norm.pdf(x, loc=mu0, scale=se)
    # Null distribution plot
    axes[0].plot(x, null_pdf, color='black', linewidth=2, label='Null distribution')
    axes[0].axvline(obs_mean, color='red', linewidth=1, linestyle = ":", label=r'Randomly sampled mean ($\bar{x})$')
    axes[0].set_ylabel("Probability density\n(area = 1)")
    # True mean line (alternative)
    axes[0].axvline(mu, color='gray', linestyle='--', label=r'Alternative mean ($\mu_1$)')
    axes[0].set_title("Sampling Distribution of the Mean", pad=48)
    #axes[0].legend(fontsize = 8)

    # Z-statistics historgram
    obs_zstat = z_stats[n]
    axes[1].hist(z_stats, bins=50, density=True, color="forestgreen", alpha=0.7, label='Simulated z-statistics')
    axes[1].axvline(obs_zstat, color='red', linewidth=1, linestyle = ":", label='Randomly sampled z-statistic')
    axes[1].axvline(z_alt, color='gray', linestyle='--', label=r'Expected $z$ under $H_1$')
    axes[1].set_title(r'Z-statistics (relative to $H_0$)', pad=48)
    #axes[1].legend(fontsize = 8)
    
    # P-values histogram
    obs_pval = p_vals[n]
    axes[2].hist(p_vals, bins=50, density=True, color="forestgreen", alpha=0.7, label='Simulated p-values')
    axes[2].axvline(0.05, color='black', linestyle='--', label='α = 0.05')
    axes[2].axvline(obs_pval, color='red', linewidth=1, linestyle = ":", label='Randomly sampled p-value')
    axes[2].set_title("P-values", pad=48)
    #axes[2].legend(fontsize=8)
    # Shade the "significant" region
    axes[2].axvspan(0, 0.05, color='lightgray', alpha=0.3, label=f'Significant area (empirical power ~ {np.mean(p_vals<0.05):.2f})')

    # Legends
    for ax in axes:
        ax.legend(
            loc='lower center',
            bbox_to_anchor=(0.5, 1.03),
            ncol=2,
            fontsize=8,
            frameon=False
    )
    
    # Power annotation
    fig.text(
        0.5, -0.05,
        f"Empirical power (P[p < 0.05 | H₁]): {power:.3f}",
        ha="center",
        fontsize=12
    )
    
    plt.tight_layout()
    plt.subplots_adjust(top=0.78)
    plt.show()
```

Correlation and <i>p</i>-values

<hr style="border:0; border-top:1px solid #999; margin:0 0 16px 0;">

<p style="font-size:16px; line-height:1.65; color:#444; margin-top:0;">
We now extend the same ideas to the relationship between two variables. Repeated simulated samples allow us to examine how the sample correlation varies, how that variability is translated into a test statistic and <i>p</i>-value, and how sample size and the strength of the underlying correlation affect statistical power.
</p>

</div>

```python
rng = default_rng()

def generate_correlated_data(n, rho):
    mean = [0, 0]
    cov = [[1, rho],
           [rho, 1]]
    x, y = rng.multivariate_normal(mean, cov, size=n).T
    return x, y
```

<div style="max-width:950px; margin:36px auto 14px auto; font-family:Georgia, 'Times New Roman', serif;">

<div style="font-family:Arial, sans-serif; font-size:12px; font-weight:600; letter-spacing:1.2px; text-transform:uppercase; color:#777; margin-bottom:4px;">
Simulation step
</div>

<h3 style="font-size:18px; font-weight:600; color:#333; margin-top:0; margin-bottom:6px;">
Generate correlated variables
</h3>

<p style="font-size:15px; line-height:1.6; color:#555; margin-top:0; margin-bottom:0;">
The code below generates repeated samples of two variables with a specified population correlation, &rho;. For each independent sample, the sample correlation, <i>r</i>, is calculated along with its corresponding test statistic and <i>p</i>-value.
</p>

</div>

```python
def correlation_sampling_and_power(rho, sample_size, n_sims):

    # Significance level
    alpha = 0.05

    # Storage
    x_example = None
    y_example = None
    r_vals = []
    t_vals = []
    p_vals = []

    # Repeated simulated samples
    # ---------------------------------------------------------
    for i in range(n_sims):

        x, y = generate_correlated_data(sample_size, rho)

        # Save the first simulated sample as one random experiment
        if i == 0:
            x_example = x.copy()
            y_example = y.copy()

            # Least-squares regression line for the example experiment
            slope, intercept = np.polyfit(x_example, y_example, 1)

            x_line = np.linspace(x_example.min(), x_example.max(), 100)
            y_line = intercept + slope * x_line

        # Sample correlation
        r = np.corrcoef(x, y)[0, 1]

        # Test statistic for H0: rho = 0
        t_stat = r * np.sqrt(
            (sample_size - 2) / (1 - r**2)
        )

        # Two-sided p-value
        p = 2 * (
            1 - t.cdf(abs(t_stat), df=sample_size - 2)
        )

        r_vals.append(r)
        t_vals.append(t_stat)
        p_vals.append(p)

    # Convert to arrays
    r_vals = np.array(r_vals)
    t_vals = np.array(t_vals)
    p_vals = np.array(p_vals)

    # Statistics describing repeated experiments
    # ---------------------------------------------------------
    r_sd = np.std(r_vals)

    # Fraction of simulated samples with p < alpha
    rejection_rate = np.mean(p_vals < alpha)

    # The first simulated sample is our "one random experiment"
    r_random = r_vals[0]
    t_random = t_vals[0]
    p_random = p_vals[0]

    # Plot
    # ---------------------------------------------------------
    fig, axes = plt.subplots(1, 4, figsize=(20, 4))

    # Panel 1: One random experiment
    # =========================================================
    axes[0].scatter(
        x_example,
        y_example,
        color="dimgray",
        alpha=0.7
    )

    axes[0].axhline(
        0,
        color="gray",
        linewidth=1
    )

    axes[0].axvline(
        0,
        color="gray",
        linewidth=1
    )

    axes[0].plot(
        x_line,
        y_line,
        color="red",
        linewidth=2
    )

    axes[0].set_title(
        "One Random Experiment",
        pad=48
    )

    axes[0].set_xlabel("X")
    axes[0].set_ylabel("Y")

    axes[0].text(
        0.05,
        0.95,
        f"y = {intercept:.2f} + {slope:.2f}x",
        transform=axes[0].transAxes,
        fontsize=10,
        verticalalignment="top"
    )

    # Panel 2: Sampling distribution of r
    # =========================================================
    axes[1].hist(
        r_vals,
        bins=40,
        density=True,
        color="forestgreen",
        alpha=0.7,
        label=r"Simulated $r$ values"
    )

    # Null correlation
    axes[1].axvline(
        0,
        color="black",
        linestyle="-",
        linewidth=1.5,
        label=r"Null correlation ($\rho_0 = 0$)"
    )

    # True population correlation
    axes[1].axvline(
        rho,
        color="gray",
        linestyle="--",
        linewidth=1.5,
        label=r"True correlation ($\rho$)"
    )

    # Correlation from the one random experiment
    axes[1].axvline(
        r_random,
        color="red",
        linestyle=":",
        linewidth=1.5,
        label=r"Randomly sampled $r$"
    )

    axes[1].set_title(
        r"Sampling Distribution of $r$",
        pad=48
    )

    axes[1].set_ylabel(
        "Probability density\n(area = 1)"
    )

    # Panel 3: t-statistics
    # =========================================================
    axes[2].hist(
        t_vals,
        bins=40,
        density=True,
        color="forestgreen",
        alpha=0.7,
        label=r"Simulated $t$-statistics"
    )

    # Null reference
    axes[2].axvline(
        0,
        color="black",
        linestyle="-",
        linewidth=1.5,
        label=r"Null ($H_0$)"
    )

    # t-statistic from the one random experiment
    axes[2].axvline(
        t_random,
        color="red",
        linestyle=":",
        linewidth=1.5,
        label=r"Randomly sampled $t$-statistic"
    )

    axes[2].set_title(
        r"$t$-statistics (relative to $H_0$)",
        pad=48
    )

    axes[2].set_ylabel(
        "Probability density\n(area = 1)"
    )

    # Panel 4: p-values
    # =========================================================
    axes[3].hist(
        p_vals,
        bins=40,
        density=True,
        color="forestgreen",
        alpha=0.7,
        label=r"Simulated $p$-values"
    )

    # Significance threshold
    axes[3].axvline(
        alpha,
        color="gray",
        linestyle="--",
        linewidth=1.5,
        label=rf"$\alpha = {alpha:.2f}$"
    )

    # p-value from the one random experiment
    axes[3].axvline(
        p_random,
        color="red",
        linestyle=":",
        linewidth=1.5,
        label=r"Randomly sampled $p$-value"
    )

    # Region where H0 would be rejected
    axes[3].axvspan(
        0,
        alpha,
        color="orange",
        alpha=0.3
    )

    axes[3].set_title(
        r"$p$-values",
        pad=48
    )

    # Legends
    # ---------------------------------------------------------
    for ax in axes[1:]:
        ax.legend(
            loc="lower center",
            bbox_to_anchor=(0.5, 1.03),
            ncol=2,
            fontsize=8,
            frameon=False
        )

    # Bottom annotation
    # ---------------------------------------------------------
    if np.isclose(rho, 0.0):

        inference_text = (
            rf"Empirical rejection rate: "
            rf"$P(p < {alpha:.2f}\mid \rho = 0) = "
            rf"{rejection_rate:.3f}$"
        )

    else:

        inference_text = (
            rf"Empirical power: "
            rf"$P(p < {alpha:.2f}\mid \rho = {rho:.2f}) = "
            rf"{rejection_rate:.3f}$"
        )

    fig.text(
        0.5,
        0.01,
        rf"Sampling SD of $r$ = {r_sd:.3f}   |   "
        + inference_text,
        ha="center",
        fontsize=10
    )

    # Leave room for legends and bottom annotation
    plt.tight_layout()
    plt.subplots_adjust(
        top=0.78,
        bottom=0.16
    )

    plt.show()
```
